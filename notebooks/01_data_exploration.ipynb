```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')
```

```python
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
sns.set_style('whitegrid')
```

```python
DATA_RAW = Path('../data/raw')
DATA_PROCESSED = Path('../data/processed')
DATA_PROCESSED.mkdir(parents=True, exist_ok=True)
```

```python
ml_ratings = pd.read_csv(
    DATA_RAW / 'ml-32m/ratings.csv',
    sep='::',
    engine='python',
    names=['user_id', 'movie_id', 'rating', 'timestamp'],
    encoding='latin-1'
)

ml_movies = pd.read_csv(
    DATA_RAW / 'ml-32m/movies.csv',
    sep='::',
    engine='python',
    names=['movie_id', 'title', 'genres'],
    encoding='latin-1'
)
```

```python
print(f"MovieLens рейтинги: {ml_ratings.shape}")
print(f"MovieLens фильмы: {ml_movies.shape}")
```

```python
print(f"\nUnique user_ids: {ml_ratings['user_id'].nunique()}")
print(f"Unique movie_ids: {ml_ratings['movie_id'].nunique()}")
print(f"Any NaN in user_id: {ml_ratings['user_id'].isna().any()}")
print(f"Any NaN in movie_id: {ml_ratings['movie_id'].isna().any()}")
print(f"Any NaN in rating: {ml_ratings['rating'].isna().any()}")
```

```python
if ml_ratings['user_id'].nunique() == 0:
    print("Warning: No unique user_ids found in ml_ratings!")
if ml_ratings['movie_id'].nunique() == 0:
    print("Warning: No unique movie_ids found in ml_ratings!")
```

```python
num_users = ml_ratings['user_id'].nunique()
num_items = ml_ratings['movie_id'].nunique()

if num_users > 0 and num_items > 0:
    density = len(ml_ratings) / (num_users * num_items)
    print(f"\nПлотность матрицы: {density:.4%}")
else:
    print(f"\nПлотность матрицы: не может быть вычислена, num_users: {num_users}, num_items: {num_items}")
```

```python
print("=== MovieLens Statistics ===")
print(f"Пользователей: {ml_ratings['user_id'].nunique():,}")
print(f"Фильмов: {ml_ratings['movie_id'].nunique():,}")
print(f"Взаимодействий: {len(ml_ratings):,}")
print(f"Плотность матрицы: {len(ml_ratings) / (ml_ratings['user_id'].nunique() * ml_ratings['movie_id'].nunique()):.4%}")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
```

```python
ml_ratings['rating'].value_counts().sort_index().plot(kind='bar', ax=axes[0])
axes[0].set_title('Распределение рейтингов')
axes[0].set_xlabel('Рейтинг')
axes[0].set_ylabel('Количество')
```

```python
user_counts = ml_ratings['user_id'].value_counts()
axes[1].hist(user_counts, bins=50, edgecolor='black')
axes[1].set_title('Рейтингов на пользователя')
axes[1].set_xlabel('Количество рейтингов')
axes[1].set_ylabel('Пользователей')
axes[1].set_yscale('log')
```

```python
movie_counts = ml_ratings['movie_id'].value_counts()
axes[2].hist(movie_counts, bins=50, edgecolor='black')
axes[2].set_title('Рейтингов на фильм')
axes[2].set_xlabel('Количество рейтингов')
axes[2].set_ylabel('Фильмов')
axes[2].set_yscale('log')

plt.tight_layout()
plt.show()
```
